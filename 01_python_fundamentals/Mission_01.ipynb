### Problem 2. Remove Vowels

* **Problem Description**: Write a function that returns a string with the vowels (a, e, i, o, u) removed from the input string.
* **Function Signature**: `remove_vowels(s: str) -> str`
* **Input/Output Examples**:
    * Input: `"codeit"` / Output: `"cdt"`
    * Input: `"python programming"` / Output: `"pythn prgrmmng"`

- **✅ Python `len()` Function Summary:**
    - The `len()` function returns the **number of items (length)** contained within a **Sequence data type** (e.g., String, List, Tuple) in Python.

- **Usage Syntax:**
    - `len(sequence_variable_or_value)`

- **How it Works (Relevant to Engineering Data):**
    - **Input:** Any sequence, such as a material code string, a list of measurements, or a file name.
    - **Output:** The **precise count of elements** within that input.
    - **Counting Rule:** `len()` accurately counts every Unicode character as a single unit, including **letters, numbers, special characters, and spaces**.

- **Example and Result:**
    - `len("KAIST 연구")`
        - **Result:** **8**
        - **Composition:** `K`, `A`, `I`, `S`, `T`, **(space)**, `연`, `구` (8 characters total)

---

````python
def remove_vowels(s):
    # 1. Define all vowels (both lowercase and uppercase) to be removed.
    VOWELS = "aeiouAEIOU"

    # 2. Initialize an empty string to store the characters that are NOT vowels.
    removed_one = ""

    # 3. Iterate through each character ('char') in the input string 's'.
    for char in s:
        # 4. Check if the current character is NOT present in the VOWELS set.
        #    This is the core filtering logic.
        if char not in VOWELS:
            # 5. If the character is a consonant (or any non-vowel), append it to the result string.
            removed_one += char

    # 6. Return the new string containing only consonants.
    return removed_one
````

### Problem 3. Compare Numbers

* **Problem Description**: Write a function that takes two numbers and returns the larger and smaller number. If the numbers are equal, return the number once.

### Problem 5. Calculate Total Sum

* **Problem Description**: Write a function that calculates the sum of all integers from 0 up to a user-provided number $n$.
* **Function Signature**: `sum_up_to_n(n) -> int`
* **Input/Output Examples**:
    * Input: `3` / Output: `6`
    * Input: `10` / Output: `55`

---
- **Range function Functionality:**
    - The `range()` function is Python's highly efficient tool used primarily in `for` loops to **generate a sequence of integers**.
    - It generates numbers **on demand**, saving memory compared to creating a full list of numbers.

- **Usage Syntax (Three Forms):**
    - **1. `range(stop)`:** Generates numbers **starting from 0** up to (but **not including**) the `stop` number.
        - *Example:* `range(5)` → 0, 1, 2, 3, 4
    - **2. `range(start, stop)`:** Generates numbers starting from `start` up to (but **not including**) the `stop` number.
        - *Example:* `range(2, 7)` → 2, 3, 4, 5, 6
    - **3. `range(start, stop, step)`:** Generates numbers from `start` to `stop` (exclusive), incrementing by the specified `step`.
        - *Example:* `range(1, 10, 2)` → 1, 3, 5, 7, 9

- **Key Rule (Crucial for Engineers):**
    - The **`stop` value is always exclusive**. If you need to include `N` in your sequence (e.g., $1$ to $N$), you must set the stop value to **`N + 1`** (e.g., `range(1, N + 1)`).

---

````python
def sum_up_to_n(n):
    sum = 0
    for i in range(1,n+1):
        sum += i
    print(sum)
````

````python
sum_up_to_n(3)
sum_up_to_n(10)
````

### Problem 6. Calculate Even Sum

* **Problem Description**: Write a function that calculates the sum of even numbers from 1 up to a user-provided number $n$. If the user enters 0, the program should terminate.
* **Function Signature**: `sum_even_numbers(n: int) -> str`
* **Input/Output Examples**:
    * Input: `9` / Output: `"1부터 9까지의 짝수 합은 20입니다."`
    * Input: `0` / Output: `"종료"`

````python
def sum_even_numbers(n):
    # Check for the exit condition (Input Validation).
    if n == 0:
        print("종료")
        return  # Exit the function if n is 0 (Implicitly returns None).

    sum = 0
    # Loop from 1 up to 'n' (inclusive).
    for i in range(1, n + 1):
        # Check if the current number 'i' is even (remainder is 0).
        if i % 2 == 0:
            # Add the even number to the total sum.
            sum += i

    # Print the final result using an f-string.
    print(f"1부터 {n}까지의 짝수 합은 {sum} 입니다")
    # Implicitly returns None.
````

````python
sum_even_numbers(9)
sum_even_numbers(0)
````

### Problem 1. String Compression Game

* **Problem Description**: Compress a given string by replacing consecutive repeating characters with the character and its count.
* **Function Signature**: `compress_string(s: str) -> str`
* **Input/Output Examples**:
    * Input: `"aaabbaaa"` / Output: `"a3b2a3"`
    * Input: `"aabcccccaaa"` / Output: `"a2b1c5a3"`

````python
def compress_string(string):
    # Set the first character as the memory for comparison.
    memory = string[0]
    output = ""
    # Start the counter for the current character at 1.
    count = 1
    leng = len(string)

    # Loop starts from the second character (index 1).
    for i in range(1, leng):
        # If the current character is the same as the memory:
        if memory == string[i]:
            count += 1 # Increment the count.

        # If the current character is different (end of a run):
        else:
            # Append the previous character and its count to the output.
            output += memory
            output += str(count)

            # Reset the count to 1 for the new character.
            count = 1
            # Update the memory to the new character.
            memory = string[i]

    # After the loop finishes, append the last remaining character and its count.
    output += memory
    output += str(count)

    print(output)
````

````python
compress_string("aaabbaaa")
compress_string("aabcccccaaa")
````

### Problem 3. Countdown Timer

* **Problem Description**: Receives time (seconds) from the user, counts down, printing the remaining time every second, and prints "Timer done!" when finished.
* **Function Signature**: `count_down(seconds: int) -> None`
* **Input/Output Examples**:
    * Input: `5` / Output: `"Timer Start!"`, `5`, `4`, `3`, `2`, `1`, `"Timer done!"`

````python
import time

def count_down(seconds: int) -> None:
    # Print the timer start message.
    print("Timer Start!")

    # Store the input seconds in the counter variable.
    current_time = seconds

    # Loop as long as the current time is greater than zero.
    while current_time > 0:
        # Print the current time remaining.
        print(current_time)

        # Pause the program execution for 1 second.
        time.sleep(1)

        # Decrement the time by 1.
        current_time -= 1

    # Print the termination message when the countdown is finished.
    print("Timer done!")
````

````python
count_down(5)
````

### Problem 1. Time Tracker Class (TimeTracker)

* **Problem Description**: Implement a `TimeTracker` class to record start time, stop time, and calculate elapsed time in **minutes**. Use the `datetime` module.
* **Methods**: `start`, `stop`, `get_elapsed_time`



````python
from datetime import datetime
from typing import Optional

class TimeTracker:
    def __init__(self):
        # Initialize start and stop time variables to None.
        self.start_time: Optional[datetime] = None
        self.stop_time: Optional[datetime] = None

    def start(self) -> None:
        """Records the current time to start the timer."""
        # Record the current system time using datetime.now().
        self.start_time = datetime.now()
        # Reset stop_time in case the timer is being restarted.
        self.stop_time = None
        print(f"Timer started at: {self.start_time.strftime('%Y-%m-%d %H:%M:%S')}")

    def stop(self) -> None:
        """Records the current time to stop the timer."""
        # Check if the timer was started before stopping.
        if self.start_time is None:
            print("Error: Timer has not been started yet.")
            return

        # Record the current time as the stop time.
        self.stop_time = datetime.now()
        print(f"Timer stopped at: {self.stop_time.strftime('%Y-%m-%d %H:%M:%S')}")

    def get_elapsed_time(self) -> Optional[float]:
        """Calculates the elapsed time between start and stop in minutes."""
        # Check for error if the timer was never started.
        if self.start_time is None:
            print("Error: Timer has not been started.")
            return None

        # If the timer is still running, calculate time up to the present moment.
        if self.stop_time is None:
            time_difference = datetime.now() - self.start_time
        # Otherwise, calculate time between recorded start and stop times.
        else:
            time_difference = self.stop_time - self.start_time

        # Convert the timedelta object to total seconds, then divide by 60 for minutes.
        elapsed_minutes = time_difference.total_seconds() / 60

        # Return the elapsed time, rounded to two decimal places.
        return round(elapsed_minutes, 2)
````

#### Problem 3. Employee Management Class

* **Problem Description**: Implement an `EmployeeManager` class using a **class variable** (`all_employees`) to store all employee data and a **class method** (`calculate_average_salary`) to calculate the average salary.
* **Methods**: `__init__`, `calculate_average_salary` (Class Method)



````python
class EmployeeManager:
    # Class variable to store all employee records (name as key)
    employee_records = {} # 딕셔너리를 전체로 불러야 언제든지 쓰일 수 있는 것들이 됨.

    # Constructor: Initializes a new employee and adds them to the class record
    def __init__(self, name, balance):
        new_employee_details = {
            "name": name,
            "balance": balance
        }
        # Prints employee's name and salary in English
        print(f"The salary of {name} is {balance}")
        # Adds the new employee's details to the class-level dictionary
        EmployeeManager.employee_records[name] = new_employee_details # 이런식으로 추가해야, 딕셔너리에 추가됨. (이게 없으면, 그냥 리스트는 누적이 아닌 항상 초기화됨.)

    # Static method: Calculates the average salary of all recorded employees
    def calculate_average_salary():
        sum_sal = 0  # Initialize total salary sum
        num = 0      # Initialize employee count

        # Iterate through all employee records stored in the class variable
        for name, details in EmployeeManager.employee_records.items():
            sum_sal += details["balance"]
            num += 1

        # Check to prevent division by zero
        if num == 0:
            return 0

        average = sum_sal / num  # Calculate the average salary

        return average
````

````python
class ReservationSystem:
    # If ReservationSystem objects are created, both objects would share the same reservation list, making per-branch management impossible.
    def __init__(self, name):
        # We cannot use ReservationSystem.restaurant here, as there is no input value to initialize with (it would always be an empty dictionary).
        self.branch_name = name # By using 'self', this variable becomes accessible by other instances.
        self.reservations = {}
        # pass is removed for efficiency.

    def add_reservation(self, name, date, ppl):
        details = {
            "date": date,
            "ppl": ppl
        }
        self.reservations[name] = details # Add reservation using customer name as key.

    def list_reservations(self):
        # NOTE: Print format is still incorrect and will cause issues.
        for name, details in self.reservations.items():
            print(f"{self.branch_name} Reservation List: \n {name}, Date : {details['date']}, Guests : {details['ppl']})")

    @staticmethod # No 'self' needed; saves unnecessary resource usage.
    def sum_reservations(restaurant_list): # Changed parameter name to 'restaurant_list' for clarity.
        total_count = 0
        for rest in restaurant_list:
            total_count += len(rest.reservations) # Accesses the instance variable set in __init__.
        return total_count
````

````python
up=input("Up to what number would you like to calculate the sum of primes")

number=list(range(2,int(up)+1))
sum = 0

for num_1 in range(2,int(up)):
    for num_2 in range(2,int(up)):
        mul = num_1*num_2

        if mul > int(up):
            break

        if mul in number:
            number.remove(mul)

for n in range(len(number)):
    sum += number[n]


print(f"The sum of primes up to 10 is {sum}")
````

**Would you like me to create the Python class structure (skeleton code) for this assignment?**

````python
class LibraryManagement:
    def __init__(self):
        # Repository for book info: { 'ISBN': {'title': title, 'author': author, 'year': year} }
        self.books = {}
        # Repository for member info: { 'Name': {'rented_books': [List of ISBNs]} }
        self.members = {}

    # 1. Add Book
    def add_book(self, title, author, year, isbn):
        if isbn in self.books:
            print(f"Error: Book with ISBN {isbn} already exists.")
            return

        # Store detailed information as a dictionary instead of an object
        self.books[isbn] = {
            'title': title,
            'author': author,
            'year': year
        }
        print(f"Book '{title}' has been added.")

    # 2. Add Member
    def add_member(self, name):
        if name in self.members:
            print(f"Error: Member '{name}' already exists.")
            return

        # Initialize member with an empty list for rented books
        self.members[name] = {'rented_books': []}
        print(f"Member '{name}' has been registered.")

    # 3. Rent Book
    def rent_book(self, isbn, name):
        # Validation: Check if book and member exist
        if isbn not in self.books:
            print("Error: Book not found.")
            return
        if name not in self.members:
            print("Error: Member not found.")
            return

        # Add ISBN to the member's rental list
        self.members[name]['rented_books'].append(isbn)

        # Retrieve title for confirmation message
        book_title = self.books[isbn]['title']
        print(f"Member '{name}' has rented '{book_title}'.")

    # 4. Return Book
    def return_book(self, isbn, name):
        if name not in self.members:
            print("Error: Member not found.")
            return

        member_rent_list = self.members[name]['rented_books']

        # Check if the member actually rented this specific book
        if isbn in member_rent_list:
            member_rent_list.remove(isbn)
            book_title = self.books[isbn]['title']
            print(f"Member '{name}' has returned '{book_title}'.")
        else:
            print(f"Error: Member has not rented book with ISBN {isbn}.")

    # 5. Print Books
    def print_books(self):
        print("\n[Book List]")
        for info in self.books.values():
            print(f"- {info['title']} (Author: {info['author']})")

    # 6. Print Members
    def print_members(self):
        print("\n[Member List]")
        for name, info in self.members.items():
            # Convert the list of ISBNs into a list of book titles for display
            rented_titles = []
            for isbn in info['rented_books']:
                rented_titles.append(self.books[isbn]['title'])

            # Format the output string
            status = ", ".join(rented_titles) if rented_titles else "None"
            print(f"- {name} (Rented: {status})")

# --- Execution Test ---
library = LibraryManagement()

# Add books
library.add_book("1984", "George Orwell", 1949, "978-001")
library.add_book("Little Prince", "Saint-Exupery", 1943, "978-002")

# Add member
library.add_member("Jae-Moon")

# Rent book
library.rent_book("978-001", "Jae-Moon")

# Print status
library.print_members()

# Return book
library.return_book("978-001", "Jae-Moon")

# Print status again
library.print_members()
````
