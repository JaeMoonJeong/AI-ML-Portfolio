```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. 결과 데이터 로드
df = pd.read_csv('outputs/result.csv')

# 2. 시각화 스타일 설정
plt.style.use('seaborn-v0_8-whitegrid')
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# [왼쪽 그래프] 공부 시간에 따른 성적 예측 분포 (산점도 + 회귀선)
sns.regplot(x='Hours Studied', y='Predicted_Performance_Index', data=df, 
            ax=axes[0], scatter_kws={'alpha':0.5}, line_kws={'color':'red'})
axes[0].set_title('Study Hours vs Predicted Performance', fontsize=14)
axes[0].set_xlabel('Hours Studied')
axes[0].set_ylabel('Predicted Index')

# [오른쪽 그래프] 이전 성적과 예측 성적의 관계
sns.scatterplot(x='Previous Scores', y='Predicted_Performance_Index', 
                hue='Extracurricular Activities', data=df, ax=axes[1])
axes[1].set_title('Previous Scores vs Predicted Performance', fontsize=14)
axes[1].set_xlabel('Previous Scores')
axes[1].set_ylabel('Predicted Index')

plt.tight_layout()
plt.show()

# 3. 주요 상관관계 수치 확인
print("\n--- 상관계수(Correlation) 분석 ---")
print(df[['Hours Studied', 'Previous Scores', 'Predicted_Performance_Index']].corr()['Predicted_Performance_Index'])
```
